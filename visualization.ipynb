```python
def add_index(sample, idx):
    sample['id'] = idx
    return sample
```

```python
dataset_caltech = load_dataset("flwrlabs/caltech101", split="train", streaming=True, trust_remote_code=True)
dataset_caltech = dataset_caltech.map(add_index, with_indices=True)
```

```python
def get_images(top_5, dataset):
  ids = set(top_5['id_1']).union(set(top_5['id_2']))
  max_id = int(max(ids))

  images = {}
  dataset_iterator = iter(dataset)

  for current_id, sample in enumerate(dataset_iterator):
      if current_id in ids:
          images[current_id] = sample['image']

      if current_id >= max_id:
          break

  return images
```

```python
def visualize(file_path, dataset, dataset_name, base_model_name):
    df = pd.read_csv(file_path, sep=";")
    top_5 = df.head(5)

    images = get_images(top_5, dataset)

    fig, axes = plt.subplots(nrows=5, ncols=2, figsize=(10, 20))
    fig.suptitle(f"Blind spots | Dataset: {dataset_name} | Base model: {base_model_name}", fontsize=16, y=0.92)

    for i, row in top_5.iterrows():
        id_1 = int(row['id_1'])
        id_2 = int(row['id_2'])

        img1 = images[id_1]
        img2 = images[id_2]

        axes[i, 0].imshow(img1)
        axes[i, 0].set_title(f"ID: {id_1}")
        axes[i, 0].axis('off')

        axes[i, 1].imshow(img2)
        axes[i, 1].set_title(f"ID: {id_2}")
        axes[i, 1].axis('off')

    plt.tight_layout()
    plt.subplots_adjust(top=0.88)
    plt.show()

    display(top_5)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

pliki = [
    "/content/vision_blindspots/clip_caltech_blind_spots.csv",
    "/content/vision_blindspots/dino_caltech_blind_spots.csv",
    "/content/vision_blindspots/siglip2_caltech_blind_spots.csv",
    "/content/vision_blindspots/siglip2_imagenet_test_blind_spots.csv",
    "/content/vision_blindspots/dino_imagenet_test_blind_spots.csv",
    "/content/vision_blindspots/clip_imagenet_test_blind_spots.csv",
    "/content/vision_blindspots/siglip_caltech_blind_spots.csv",
    "/content/vision_blindspots/siglip_imagenet_test_blind_spots.csv"
]

dane = []
for plik in pliki:
    if os.path.exists(plik):
        df_temp = pd.read_csv(plik, sep=";")
        if "caltech" in plik.lower():
            df_temp['dataset'] = 'Caltech101'
        else:
            df_temp['dataset'] = 'ImageNet-test'
        dane.append(df_temp)

df = pd.concat(dane, ignore_index=True)
sns.set_theme(style="whitegrid")
THRESHOLD_DIFF = 0.6

print(f"Wczytano {len(df)} rekordów łącznie.")
```

```python
plt.figure(figsize=(10, 6))
counts = df.groupby(['base_name', 'dataset']).size().reset_index(name='count')

ax = sns.barplot(data=counts, x='base_name', y='count', hue='dataset', palette='viridis')

for container in ax.containers:
    ax.bar_label(container, padding=3)

plt.title("Liczba wykrytych  blind spotów", fontsize=14)
plt.ylabel("Liczba par zdjęć", fontsize=12)
plt.xlabel("Model bazowy", fontsize=12)
plt.legend(title='Zbiór danych')
plt.tight_layout()
plt.show()
```

```python
disagreement_records = []
for _, row in df.iterrows():
    disagreement_records.append({
        'Base': row['base_name'], 'Dataset': row['dataset'],
        'Reference': row['ref_1_name'], 'Disagrees': row['score_ref_1'] < THRESHOLD_DIFF
    })
    disagreement_records.append({
        'Base': row['base_name'], 'Dataset': row['dataset'],
        'Reference': row['ref_2_name'], 'Disagrees': row['score_ref_2'] < THRESHOLD_DIFF
    })

df_dis = pd.DataFrame(disagreement_records)
rates = df_dis.groupby(['Base', 'Dataset', 'Reference'])['Disagrees'].mean().reset_index()
rates['Disagrees'] *= 100

for dataset_name in rates['Dataset'].unique():
    data_subset = rates[rates['Dataset'] == dataset_name]
    pivot_rates = data_subset.pivot_table(index='Base', columns='Reference', values='Disagrees')

    plt.figure(figsize=(8, 4))
    sns.heatmap(pivot_rates, annot=True, fmt=".1f", cmap="YlOrRd", vmin=0, vmax=100)
    plt.title(f"% niezgodności modeli referencyjnych na zbiorze {dataset_name}", fontsize=14)
    plt.ylabel("Model Bazowy", fontsize=12)
    plt.xlabel("Model Referencyjny", fontsize=12)
    plt.tight_layout()
    plt.show()
```

```python
melt_1 = df[['base_name', 'dataset', 'ref_1_name', 'score_ref_1']].rename(
    columns={'ref_1_name': 'Ref_Model', 'score_ref_1': 'Score'})
melt_2 = df[['base_name', 'dataset', 'ref_2_name', 'score_ref_2']].rename(
    columns={'ref_2_name': 'Ref_Model', 'score_ref_2': 'Score'})
df_melted = pd.concat([melt_1, melt_2])

for base_model in df_melted['base_name'].unique():
    data_subset = df_melted[df_melted['base_name'] == base_model]

    plt.figure(figsize=(8, 5))
    sns.boxplot(data=data_subset, x='dataset', y='Score', hue='Ref_Model', palette='Set3')
    plt.axhline(THRESHOLD_DIFF, color='red', linestyle='--', label='Próg (0.6)')

    plt.title(f"Oceny modeli referencyjnych dla bazowego modelu: {base_model}", fontsize=14)
    plt.ylabel("Podobieństwo cosinusowe", fontsize=12)
    plt.xlabel("Zbiór danych", fontsize=12)
    plt.legend(title="Model referencyjny", bbox_to_anchor=(1.05, 1), loc='upper left')
    plt.tight_layout()
    plt.show()
```
